# Mulliken populations and τ1 diagnostics

Reproduces paper Figures 5–8.

Inputs: `./data/mulliken_results_with_IBM_scf_checkfiles.txt` (gas-phase Mulliken charges of the CO2 and Co(H2O)5 fragments from the (RO)HF 1-RDM; column legend at the bottom of the file) and `./data/t1_diagnostics_Co_CO2_5H2O.txt` (τ1 values; column legend at the bottom of the file). Output: `./figures/`.

## Mulliken populations

### Charge 2+: doublet and quartet (paper Figure 8)

```python
import numpy as np
import matplotlib.pyplot as plt
# Load the data (skip header if needed)
path_to_data = "./data/mulliken_results_with_IBM_scf_checkfiles.txt"
data = np.loadtxt(f"{path_to_data}", comments="#")

distances = data[:, 0]
CO2_charge_rohf_quintet = data[:, 1]
Co_5H2O_charge_rohf_quintet = data[:, 2]

CO2_charge_rohf_quartet = data[:, 3]
Co_5H2O_charge_rohf_quartet = data[:, 4]


# Plot
plt.figure(figsize=(6, 5))
plt.plot(distances, CO2_charge_rohf_quintet, label=r"$[\mathrm{S=1/2 - CO_2}]$", marker='o')
plt.plot(distances, Co_5H2O_charge_rohf_quintet, label=r"$[\mathrm{S=1/2 - Co(H_2O)_5}]$", marker='o')
plt.plot(distances, CO2_charge_rohf_quartet, label=r"$[\mathrm{S=3/2 - CO_2}]$", marker='d')
plt.plot(distances, Co_5H2O_charge_rohf_quartet, label=r"$[\mathrm{S=3/2 - Co(H_2O)_5}]$", marker='d')

plt.xlabel("Distance [$\mathrm{\\AA}$]",fontsize=14)
plt.ylabel("Mulliken Charge / e",fontsize=14)
plt.title(r"$[\mathrm{Co(H_2O)_5CO_2}]^{2+}$",fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(fontsize=10)
plt.grid(True)
plt.tight_layout()
plt.savefig('./figures/Figure_8.pdf')
plt.show()
```

### Charge 3+: singlet and quintet (paper Figure 7)

```python
import numpy as np
import matplotlib.pyplot as plt
# Load the data (skip header if needed)
path_to_data = "./data/mulliken_results_with_IBM_scf_checkfiles.txt"
data = np.loadtxt(f"{path_to_data}", comments="#")

distances = data[:, 0]
CO2_charge_rohf_quintet = data[:, 5]
Co_5H2O_charge_rohf_quintet = data[:, 6]

CO2_charge_rhf_singlet = data[:, 7]
Co_5H2O_charge_rhf_singlet = data[:, 8]


# Plot
plt.figure(figsize=(6, 5))
plt.plot(distances, CO2_charge_rohf_quintet, label=r"$[\mathrm{S=2 - CO_2}]$", marker='o')
plt.plot(distances, Co_5H2O_charge_rohf_quintet, label=r"$[\mathrm{S=2 - Co(H_2O)_5}]$", marker='o')
plt.plot(distances, CO2_charge_rhf_singlet, label=r"$[\mathrm{S=0 - CO_2}]$", marker='s')
plt.plot(distances, Co_5H2O_charge_rhf_singlet, label=r"$[\mathrm{S=0 - Co(H_2O)_5}]$", marker='s')

plt.xlabel("Distance [$\mathrm{\\AA}$]",fontsize=14)
plt.ylabel("Mulliken Charge / e",fontsize=14)
plt.title(r"$[\mathrm{Co(H_2O)_5CO_2}]^{3+}$",fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(fontsize=10)
plt.grid(True)
plt.tight_layout()
plt.savefig('./figures/Figure_7.pdf')
plt.show()
```

# t1-diagnostics

τ1 diagnostics from (U)CCSD in the AVAS active space: (30e, 25o) for charge 3+, (23e, 17o) for charge 2+.

### Charge 3+: singlet and quintet (paper Figure 5)

```python
import numpy as np
import matplotlib.pyplot as plt
# Load the data (skip header if needed)
path_to_data = "./data/t1_diagnostics_Co_CO2_5H2O.txt"
data = np.loadtxt(f"{path_to_data}", comments="#")

distances = data[:, 0]
t1_singlet_gas = data[:, 1]
t1_quintet_gas = data[:, 2]
t1_singlet_solvent = data[:, 3]
t1_quintet_solvent = data[:, 4]
t1_doublet_gas = data[:, 5]
t1_quartet_gas = data[:, 6]
t1_doublet_solvent = data[:, 7]
t1_quartet_solvent = data[:, 8]


# Plot
plt.figure(figsize=(6, 5))
plt.plot(distances, t1_singlet_gas, label="S=0, T1 diagnostic - gas phase", marker='o' )
plt.plot(distances, t1_quintet_gas, label="S=2, T1 diagnostic - gas phase", marker='o' )
plt.plot(distances, t1_singlet_solvent, label="S=0, T1 diagnostic - solvent", marker='d' )
plt.plot(distances, t1_quintet_solvent, label="S=2, T1 diagnostic - solvent", marker='d' )
# plt.plot(distances, t1_doublet_gas, label="S=1/2, T1 diagnostic - gas phase", marker='o' )
# plt.plot(distances, t1_quartet_gas, label="S=3/2, T1 diagnostic - gas phase", marker='o' )
# plt.plot(distances, t1_doublet_solvent, label="S=1/2, T1 diagnostic - solvent", marker='d' )
# plt.plot(distances, t1_quartet_solvent, label="S=3/2, T1 diagnostic - solvent", marker='d' )


# Threshold lines
# plt.axhline(0.02, color="gray", linestyle="--", linewidth=1,label=r"$\tau_1 = 0.02$")
plt.axhline(0.05, color="red", linestyle="--", linewidth=1,label=r"$\tau_1 = 0.05$")


plt.xlabel("Distance [$\mathrm{\\AA}$]",fontsize=14)
plt.ylabel(r"$\mathrm{\tau_1}$",fontsize=14)
plt.title("T1-Diagnostic",fontsize=14)
plt.ticklabel_format(axis='y', style='sci', scilimits=(0,0))
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(fontsize=10)
plt.grid(True)
plt.tight_layout()
plt.savefig('./figures/Figure_5.pdf')
plt.show()
```
